```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/kaggle/input/insurance/insurance.csv")
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import PowerTransformer
```

```python
df.head()
```

```python
categorical_cols_Ordinal = ['smoker']
categorical_cols_Nominal = ['sex', 'region']
numeric_cols = ['age', 'bmi', 'children']
```

```python
numeric_pipe = Pipeline([
    ('transformer', PowerTransformer())
])
```

```python
categorical_pipe_Nominal = Pipeline([
    ('one hot', OneHotEncoder(sparse_output=False, drop = 'first')),
])
```

```python
categorial_pipe_Ordinal = Pipeline([
    ('Ordinal', OrdinalEncoder(categories = [['yes', 'no']]))
])
```

```python
preprocessor = ColumnTransformer([
    ('Ncat', categorical_pipe_Nominal, categorical_cols_Nominal),
    ('Ocat', categorial_pipe_Ordinal, categorical_cols_Ordinal),
    ('numeric', numeric_pipe, numeric_cols),
], remainder = 'passthrough')
```

```python
pipe = Pipeline([
    ('Preprocessing', preprocessor),
    ('training', LinearRegression())
])
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns = ['charges']), df['charges'], test_size = 0.25)
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

print("\nTEST")
print("MAE :", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R²  :", r2_score(y_test, y_pred))
```
