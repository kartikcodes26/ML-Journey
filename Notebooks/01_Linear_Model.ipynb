```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([6, 7, 1, 9, 0, 2])
y = np.array([7, 8, 2, 10, 1, 3])
```

```python
X = x.reshape(-1, 1)
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
y_pred = model.predict(X)
```

```python
sns.scatterplot(
    x = x,
    y = y
)

# Regression line
sns.regplot(
    x = X,
    y = y_pred
)

plt.show()
```

```python
x_new = np.array([[20]]) # Make sure it is 2D
print(model.predict(x_new))
```
