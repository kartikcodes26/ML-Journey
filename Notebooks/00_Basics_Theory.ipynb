# Machine Learning Paradigms — Quick Revision

## 1. Supervised Learning
- Training data contains **inputs (X) + known targets (y)**.
- Model learns `X → y`.
- Goal: make accurate predictions on unseen data.
- Examples: Regression, Classification.
- Example: `Hours Studied → Exam Score`.

---

## 2. Unsupervised Learning
- Training data has **no target labels**.
- Model tries to discover hidden structure/patterns in the data.
- Examples: Clustering, dimensionality reduction, anomaly detection.
- Example: Group customers based on purchasing behavior.

---

## 3. Clustering
- An **unsupervised learning** technique.
- Groups similar data points together.
- No predefined labels are provided.
- Examples: K-Means, DBSCAN, Hierarchical Clustering.
- Example: Automatically grouping customers into different segments.

---

## 4. Semi-Supervised Learning
- Uses a **small amount of labeled data + large amount of unlabeled data**.
- Useful when labeling data is expensive.
- Example:
  - 1,000 labeled images
  - 100,000 unlabeled images
- Model uses both to improve learning.

---

## 5. Anomaly Detection
- Identifies data points that are **significantly different from normal behavior**.
- Can be:
  - Unsupervised → no labels.
  - Semi-supervised → train mainly/only on normal examples.
  - Supervised → normal + anomaly labels available.
- Example: Fraud detection, machine failure detection.

### Key idea:
> Learn what "normal" looks like → flag unusual observations.

---

## 6. Self-Supervised Learning
- Model creates **supervision from the data itself**.
- No human-provided labels are required.
- A part of the data is hidden/corrupted and the model learns to predict it.
- Commonly used for pretraining large language models and vision models, usually followed by fine-tuning.

Example:
```text
Input:  "The cat sat on the ___"
Target: "mat"
```

---

## 7. Data Masking
- Hide/mask part of the input and make the model predict the missing information.
- A common technique used *within* self-supervised learning.

Example:
```text
"The cat [MASK] on the mat"
              ↓
          predict "sat"
```

> **Note:** Masking is a *technique*; self-supervised learning is the broader *paradigm* it belongs to.

---

## 8. Transfer Learning
- Take knowledge learned by a model on one task/dataset and reuse it for another related task.
- Typical flow:
  1. Start with a pretrained model.
  2. Reuse its learned representations.
  3. Fine-tune it on the new task.

Example:
```text
ImageNet-trained model
        ↓
Reuse learned visual features
        ↓
Train for medical-image classification
```

**Benefits:**
- Requires less data.
- Requires less training time.
- Often gives better performance.

---

## 9. Reinforcement Learning (RL)
- An **agent** interacts with an **environment**.
- Agent takes **actions** and receives **rewards/penalties**.
- Learns a **policy** for maximizing cumulative reward.

```text
Agent
  ↓ action
Environment
  ↓
Reward
  ↓
Agent learns
```

| Term | Meaning |
|---|---|
| **Policy** | Strategy that determines which action to take in a given state. |
| **Reward** | Positive feedback for desirable outcomes; RL tries to maximize cumulative reward. |
| **Penalty** | Negative feedback for undesirable actions/outcomes; discourages bad behavior. |

---

## 10. Batch vs Online Learning

### Batch Learning
- Train using a fixed dataset.
- Model is trained on available data and deployed.
- New data usually requires re-training from scratch.

```text
3000 samples
     ↓
   Train
     ↓
   Model
```

### Online Learning
- Model is updated **incrementally** as new data arrives.

```text
3000 samples → Train
                  ↓
             New 150 samples
                  ↓
             Update model
                  ↓
             New model
```

- **Advantage:** Adapts to changing data.
- **Risk:** Training only on new data can make the model forget older patterns (catastrophic forgetting).

---

## 11. Data Drift vs Concept Drift

### Data Drift
- The **input** data distribution changes over time.

```text
Training (2026):  Age distribution → mostly 20–40
Production (2030): Age distribution → mostly 40–60
```
- Model may perform worse because the input distribution changed.

### Concept Drift
- The **relationship between input and target** changes over time.

```text
2026: Level characteristics → Demon difficulty
2030: The community's difficulty standards change
```

### Key distinction
| Type | What changes |
|---|---|
| Data Drift | P(X) |
| Concept Drift | P(Y \| X) |

---

## One-line Revision

- **Supervised:** Learn from X + y.
- **Unsupervised:** Find structure without y.
- **Clustering:** Group similar points.
- **Semi-supervised:** Small labeled + lots of unlabeled data.
- **Anomaly detection:** Find unusual observations.
- **Self-supervised:** Generate the learning signal from the data itself.
- **Data masking:** Hide part of input and predict it.
- **Transfer learning:** Reuse knowledge from a previous task.
- **RL:** Learn actions through rewards/penalties.
- **Policy:** Strategy for choosing actions.
- **Batch:** Train on a fixed dataset.
- **Online:** Continuously update with incoming data.
- **Data drift:** Input distribution changes.
- **Concept drift:** Input-target relationship changes.

# ML Chapter 1 — Questions & Short Answers

### 1. How would you define Machine Learning?
Machine Learning is the field of study that gives computers the ability to learn from data and improve their performance on a task without being explicitly programmed for every case.

---

### 2. Name four types of applications where ML shines.
- Problems where traditional programming rules are difficult to write
- Problems requiring adaptation to changing data
- Problems involving very large amounts of data
- Problems where humans can recognize patterns but struggle to describe them explicitly

---

### 3. What is a labeled training set?
A training dataset where each input example is accompanied by its correct target/output (label).

Example:
`image → "cat"`

---

### 4. What are the two most common supervised tasks?
- **Classification** → predict a category
- **Regression** → predict a numerical value

---

### 5. Name four common unsupervised tasks.
- **Clustering** → group similar instances
- **Dimensionality reduction** → reduce number of features
- **Anomaly detection** → detect unusual instances
- **Association rule learning** → discover relationships between features

---

### 6. What type of algorithm would you use to allow a robot to walk in unknown terrains?
**Reinforcement Learning (RL)** — the robot learns through rewards and penalties from interacting with its environment.

---

### 7. What type of algorithm would you use to segment customers into multiple groups?
**Unsupervised learning**, typically **clustering**.

---

### 8. Is spam detection supervised or unsupervised?
Usually **supervised learning**, because emails can be labeled:
`spam` or `not spam`.

---

### 9. What is an online learning system?
A system that learns **incrementally**, updating the model as new data arrives instead of training only once on a fixed dataset.

---

### 10. What is out-of-core learning?
Training a model on a dataset that **doesn't fit into memory** by processing the data in smaller chunks.

---

### 11. What type of algorithm relies on a similarity measure to make predictions?
**Instance-based learning algorithms.**

Example: **k-Nearest Neighbors (k-NN)**.

They make predictions based on how similar a new instance is to previously seen instances.

---

### 12. Difference between a model parameter and a model hyperparameter?

**Parameter:** Learned automatically from training data.

Examples:
- Weights
- Biases

**Hyperparameter:** Set by the user/algorithm designer and controls the learning process or model structure.

Examples:
- Learning rate
- Regularization strength
- Number of trees

> Parameter → **learned**
>
> Hyperparameter → **chosen**

---

### 13. What do model-based algorithms search for? How do they succeed? How do they predict?

They search for **model parameters** that make the model fit the training data well.

Most commonly, they:
1. Define a **cost/loss function**
2. Find parameters that **minimize the cost**
3. Use the learned model to make predictions on new data

A common optimization strategy is **Gradient Descent**.

---

### 14. Name four main challenges in ML.
- **Insufficient training data**
- **Nonrepresentative training data**
- **Poor-quality data**
- **Overfitting**

Other challenges include underfitting and irrelevant features.

---

### 15. Great training performance but poor generalization — what's happening? Give three solutions.
This is **overfitting**.

Possible solutions:
- Get **more training data**
- Use a **simpler model**
- Add **regularization**

---

### 16. What is a test set, and why use it?
A **test set** is data kept separate from training and model development.

It is used to get a final estimate of how well the model will perform on **unseen data**.

> Test set = **final exam**

---

### 17. What is the purpose of a validation set?
To evaluate different models and **tune hyperparameters** during development.

> Validation/dev set = **make decisions**

---

### 18. What is a train-dev set, when do you need it, and how do you use it?
A **train-dev set** is data from the same distribution as the training data, but the model does not train on it.

It is useful when you suspect a **distribution mismatch** between training and validation data.

Compare:

`Training error → Train-dev error → Validation error`

- Large Training → Train-dev gap → **Overfitting**
- Small Training → Train-dev gap but large Train-dev → Validation gap → **Distribution mismatch**

---

### 19. What can go wrong if you tune hyperparameters using the test set?
You can **overfit to the test set**.

The test set is no longer an unbiased measure of generalization because your decisions were influenced by its results.

> Training → **Learn**
>
> Validation/Dev → **Choose**
>
> Test → **Final evaluation**
>
> Train-dev → **Diagnose**
