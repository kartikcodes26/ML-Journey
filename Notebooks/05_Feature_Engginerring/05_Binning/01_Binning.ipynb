```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)

n = 1000

df = pd.DataFrame({
    "age": np.random.randint(18, 70, n),
    "income": np.random.normal(60000, 20000, n).clip(15000, 150000),
    "credit_score": np.random.normal(680, 80, n).clip(300, 850),
    "annual_spending": np.random.normal(50000, 18000, n).clip(5000, 120000)
})

df.head()
```

```python
from sklearn.preprocessing import KBinsDiscretizer
```

```python
bin1 = KBinsDiscretizer(n_bins = 10, strategy='uniform', encode = 'ordinal')
bin2 = KBinsDiscretizer(n_bins = 10, strategy='kmeans', encode = 'ordinal')
bin3 = KBinsDiscretizer(n_bins = 10, strategy='quantile', encode = 'ordinal')
```

```python
df['age_binned'] = bin2.fit_transform(df[['age']])
```

```python
df.head()
```

```python
sns.countplot(
    data = df,
    x = "age_binned",
    palette = "viridis"
)

plt.show()
```
