```python
import numpy as np
import pandas as pd

np.random.seed(42)

# 1 year of dates
dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    freq="D"
)

n = len(dates)

df = pd.DataFrame({
    "Date": dates
})


# ---------------------------------------
# Generate realistic daily expenses
# ---------------------------------------

# Base daily spending
amount = np.random.gamma(
    shape=2.5,
    scale=350,
    size=n
)

# Weekend spending tends to be higher
is_weekend = df["Date"].dt.dayofweek >= 5

amount[is_weekend] *= np.random.uniform(
    1.1, 1.5, is_weekend.sum()
)


# ---------------------------------------
# Add some random large expenses
# ---------------------------------------

large_expense_days = np.random.choice(
    n,
    size=15,
    replace=False
)

amount[large_expense_days] += np.random.randint(
    2000,
    10000,
    size=len(large_expense_days)
)


# ---------------------------------------
# Add monthly fixed expense
# ---------------------------------------

for month in range(1, 13):

    month_indices = df.index[
        df["Date"].dt.month == month
    ]

    # Random day within the month
    idx = np.random.choice(month_indices)

    amount[idx] += np.random.randint(
        8000,
        15000
    )


# Round amounts
df["Amount"] = np.round(amount, 2)


# ---------------------------------------
# Shuffle rows
# ---------------------------------------

df = df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
import pandas as pd


def add_date_features(df, col) -> pd.DataFrame:
    df = df.copy()

    date_col = df[col]

    df["Is_Weekend"] = date_col.dt.dayofweek >= 5
    df["Is_Weekday"] = date_col.dt.dayofweek < 5
    df["Year"] = date_col.dt.year
    df["Month"] = date_col.dt.month
    df["Month_Name"] = date_col.dt.month_name()
    df["Day"] = date_col.dt.day
    df["Day_Name"] = date_col.dt.day_name()
    df["Quarter"] = date_col.dt.quarter
    df["Week"] = date_col.dt.isocalendar().week

    return df
```
