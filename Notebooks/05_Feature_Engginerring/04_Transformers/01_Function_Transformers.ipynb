```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns

np.random.seed(42)

# Right-skewed / non-normal distribution
data = np.random.exponential(
    scale=50,
    size=1000
)
```

```python
sns.kdeplot(
    x = data
)

plt.show()
```

```python
stats.probplot(
    data,
    dist = "norm",
    plot = plt
)

plt.show()
```

```python
# Log transformation
from sklearn.preprocessing import FunctionTransformer

transformer = FunctionTransformer(func = np.log1p)
# np.square
# np.sqrt
# log1p = log(1 + x)
```

```python
data_normal = transformer.fit_transform(data)
```

```python
sns.kdeplot(
    data = data_normal
)

plt.show()
```

```python
stats.probplot(
    data_normal,
    dist = "norm",
    plot = plt
)

plt.show()
```

```python
from scipy.stats import skew
print(skew(data_normal))

# Learn whats best from trial and error
```

```python
transformer2 = FunctionTransformer(lambda x: x ** 2 + 2 * x + 3)
data2 = transformer2.fit_transform(data)
```

```python
stats.probplot(
    data2,
    dist = "norm",
    plot = plt
)

plt.show()
```

```python
skew(data2)
# Distribution fix → Transformation → Scale → Model
```
