# Machine Learning Development Cycle

## Step 1 : Framing the problem

### ask:
1) What is our end product
2) What problem are we solving
3) Is accurate data needed or just accurate categories
4) How big of a team is needed
5) Which models will be most effective
6) Types of learning
7) Who will use our product
8) Deadline
and so on......

## Step 2 : Gather the data
### Get the data from
1) API's
2) Raw csv files
3) Webscraping
4) Json files
5) DataBase
6) Spark 
.... etc

## Step 3 : Data-Preprocessing
### Get the data clean and ready for training
1) Remove Duplicates
2) Fix Formats
3) Remove outliers
4) Fill or remove missing data
... etc

## Step 4 : EDA
### Analyse data using EDA techniques and principles (Already done in detail in the stats course)

## Step 5 : Feature Engginerring
### Study, Add or Remove Columns
1) Take useless columns out of consideration
2) Combine columns to form a single new columns
... etc

## Step 6 : Train the model
### Involves
1) Training the Model
2) Calculating Error
3) Selecting best model
4) Training multiple models
5) Prevent overfitting
... etc

## Step 7 : Deploy the model
### Make the models outcomes easy to access and use for user's using front-end

## Step 8 : Beta Testing
### Select Quality Beta Testers, get feedback, fix bugs and improve app

## Step 9 : Optimise model
### Use less processing, Retrain model etc
