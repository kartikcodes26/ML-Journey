# Machine Learning Paradigms — Quick Revision

## 1. Supervised Learning
- Training data contains **inputs (X) + known targets (y)**.
- Model learns `X → y`.
- Goal: make accurate predictions on unseen data.
- Examples: Regression, Classification.
- Example: `Hours Studied → Exam Score`.

---

## 2. Unsupervised Learning
- Training data has **no target labels**.
- Model tries to discover hidden structure/patterns in the data.
- Examples: Clustering, dimensionality reduction, anomaly detection.
- Example: Group customers based on purchasing behavior.

---

## 3. Clustering
- An **unsupervised learning** technique.
- Groups similar data points together.
- No predefined labels are provided.
- Examples: K-Means, DBSCAN, Hierarchical Clustering.
- Example: Automatically grouping customers into different segments.

---

## 4. Semi-Supervised Learning
- Uses a **small amount of labeled data + large amount of unlabeled data**.
- Useful when labeling data is expensive.
- Example:
  - 1,000 labeled images
  - 100,000 unlabeled images
- Model uses both to improve learning.

---

## 5. Anomaly Detection
- Identifies data points that are **significantly different from normal behavior**.
- Can be:
  - Unsupervised → no labels.
  - Semi-supervised → train mainly/only on normal examples.
  - Supervised → normal + anomaly labels available.
- Example: Fraud detection, machine failure detection.

### Key idea:
> Learn what "normal" looks like → flag unusual observations.

---

## 6. Self-Supervised Learning
- Model creates **supervision from the data itself**.
- No human-provided labels are required.
- A part of the data is hidden/corrupted and the model learns to predict it.
- Commonly used for pretraining large language models and vision models, usually followed by fine-tuning.

Example:
```text
Input:  "The cat sat on the ___"
Target: "mat"
```

---

## 7. Data Masking
- Hide/mask part of the input and make the model predict the missing information.
- A common technique used *within* self-supervised learning.

Example:
```text
"The cat [MASK] on the mat"
              ↓
          predict "sat"
```

> **Note:** Masking is a *technique*; self-supervised learning is the broader *paradigm* it belongs to.

---

## 8. Transfer Learning
- Take knowledge learned by a model on one task/dataset and reuse it for another related task.
- Typical flow:
  1. Start with a pretrained model.
  2. Reuse its learned representations.
  3. Fine-tune it on the new task.

Example:
```text
ImageNet-trained model
        ↓
Reuse learned visual features
        ↓
Train for medical-image classification
```

**Benefits:**
- Requires less data.
- Requires less training time.
- Often gives better performance.

---

## 9. Reinforcement Learning (RL)
- An **agent** interacts with an **environment**.
- Agent takes **actions** and receives **rewards/penalties**.
- Learns a **policy** for maximizing cumulative reward.

```text
Agent
  ↓ action
Environment
  ↓
Reward
  ↓
Agent learns
```

| Term | Meaning |
|---|---|
| **Policy** | Strategy that determines which action to take in a given state. |
| **Reward** | Positive feedback for desirable outcomes; RL tries to maximize cumulative reward. |
| **Penalty** | Negative feedback for undesirable actions/outcomes; discourages bad behavior. |

---

## 10. Batch vs Online Learning

### Batch Learning
- Train using a fixed dataset.
- Model is trained on available data and deployed.
- New data usually requires re-training from scratch.

```text
3000 samples
     ↓
   Train
     ↓
   Model
```

### Online Learning
- Model is updated **incrementally** as new data arrives.

```text
3000 samples → Train
                  ↓
             New 150 samples
                  ↓
             Update model
                  ↓
             New model
```

- **Advantage:** Adapts to changing data.
- **Risk:** Training only on new data can make the model forget older patterns (catastrophic forgetting).

---

## 11. Data Drift vs Concept Drift

### Data Drift
- The **input** data distribution changes over time.

```text
Training (2026):  Age distribution → mostly 20–40
Production (2030): Age distribution → mostly 40–60
```
- Model may perform worse because the input distribution changed.

### Concept Drift
- The **relationship between input and target** changes over time.

```text
2026: Level characteristics → Demon difficulty
2030: The community's difficulty standards change
```

### Key distinction
| Type | What changes |
|---|---|
| Data Drift | P(X) |
| Concept Drift | P(Y \| X) |

## One-line Revision

- **Supervised:** Learn from X + y.
- **Unsupervised:** Find structure without y.
- **Clustering:** Group similar points.
- **Semi-supervised:** Small labeled + lots of unlabeled data.
- **Anomaly detection:** Find unusual observations.
- **Self-supervised:** Generate the learning signal from the data itself.
- **Data masking:** Hide part of input and predict it.
- **Transfer learning:** Reuse knowledge from a previous task.
- **RL:** Learn actions through rewards/penalties.
- **Policy:** Strategy for choosing actions.
- **Batch:** Train on a fixed dataset.
- **Online:** Continuously update with incoming data.
- **Data drift:** Input distribution changes.
- **Concept drift:** Input-target relationship changes.
